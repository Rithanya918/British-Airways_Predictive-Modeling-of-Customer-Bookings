````python
import pandas as pd

# Calculate total capacity for each flight
df['TOTAL_CAPACITY'] = df['FIRST_CLASS_SEATS'] + df['BUSINESS_CLASS_SEATS'] + df['ECONOMY_SEATS']

# Calculate total passengers for each flight
df['TOTAL_PASSENGERS'] = df['TIER1_ELIGIBLE_PAX'] + df['TIER2_ELIGIBLE_PAX'] + df['TIER3_ELIGIBLE_PAX']

# Combine FLIGHT_DATE and FLIGHT_TIME into a single datetime column
df['SCHEDULED_DEPARTURE'] = pd.to_datetime(
    df['FLIGHT_DATE'].astype(str) + ' ' + df['FLIGHT_TIME'].astype(str)
)

# Create TIME_CATEGORY from SCHEDULED_DEPARTURE before deleting TIME_OF_DAY
df['TIME_CATEGORY'] = pd.cut(
    df['SCHEDULED_DEPARTURE'].dt.hour,
    bins=[0, 6, 12, 18, 24],
    labels=['Night', 'Morning', 'Afternoon', 'Evening'],
    include_lowest=True
)

# Display the results
print("First few rows with new columns:")
print(df[['FLIGHT_DATE', 'FLIGHT_TIME', 'SCHEDULED_DEPARTURE',
          'FIRST_CLASS_SEATS', 'BUSINESS_CLASS_SEATS', 'ECONOMY_SEATS', 'TOTAL_CAPACITY',
          'TIER1_ELIGIBLE_PAX', 'TIER2_ELIGIBLE_PAX', 'TIER3_ELIGIBLE_PAX', 'TOTAL_PASSENGERS']].head(10))

# Verify the new columns
print("\n\nDataset Info after adding new columns:")
print(df.info())

# Summary statistics for the new columns
print("\n\nSummary Statistics:")
print(df[['TOTAL_CAPACITY', 'TOTAL_PASSENGERS']].describe())

# Check for any issues with the datetime conversion
print(f"\n\nDate range: {df['SCHEDULED_DEPARTURE'].min()} to {df['SCHEDULED_DEPARTURE'].max()}")
print(f"Any null values in SCHEDULED_DEPARTURE: {df['SCHEDULED_DEPARTURE'].isnull().sum()}")
````

````python
# Define the time categorization function
def get_time_category(hour):
    if 6 <= hour <= 10:
        return 'Peak Business (PB)'  # 06:00 to 10:00
    elif 10 < hour <= 16:
        return 'Off-Peak/Leisure (OP)'  # 10:01 to 16:00
    elif 16 < hour <= 22:
        return 'Evening/Sleeper (ES)'  # 16:01 to 22:00
    else:
        return 'Night/Undefined'  # Outside 06:00-22:00 window

# Apply the function to create TIME_CATEGORY column
df['TIME_CATEGORY'] = df['SCHEDULED_DEPARTURE'].dt.hour.apply(get_time_category)

# Display the results
print("Time Category Distribution:")
print(df['TIME_CATEGORY'].value_counts().sort_index())

print("\n" + "="*70)
print("\nSample data with TIME_CATEGORY:")
print(df[['SCHEDULED_DEPARTURE', 'TIME_CATEGORY']].head(15))

# Verify the categorization
print("\n" + "="*70)
print("\nTime Category Breakdown:")
for category in df['TIME_CATEGORY'].unique():
    count = len(df[df['TIME_CATEGORY'] == category])
    percentage = (count / len(df)) * 100
    print(f"{category}: {count} flights ({percentage:.2f}%)")
````

- Sum up estimated demand for each tier

````python
import pandas as pd

# Step 1: Reload the data fresh (to avoid any previous issues)
file_path = '/content/drive/My Drive/Colab Notebooks/British airways/British Airways Task 1.xlsx'
df = pd.read_excel(file_path)

print(f"✅ Data loaded: {len(df)} flights")

# Step 2: Recreate all necessary features
# Calculate totals
df['TOTAL_CAPACITY'] = df['FIRST_CLASS_SEATS'] + df['BUSINESS_CLASS_SEATS'] + df['ECONOMY_SEATS']
df['TOTAL_PASSENGERS'] = df['TIER1_ELIGIBLE_PAX'] + df['TIER2_ELIGIBLE_PAX'] + df['TIER3_ELIGIBLE_PAX']

# Create scheduled departure
df['SCHEDULED_DEPARTURE'] = pd.to_datetime(
    df['FLIGHT_DATE'].astype(str) + ' ' + df['FLIGHT_TIME'].astype(str)
)

# Define time categorization function
def get_time_category(hour):
    if 6 <= hour <= 10:
        return 'Peak Business (PB)'
    elif 10 < hour <= 16:
        return 'Off-Peak/Leisure (OP)'
    elif 16 < hour <= 22:
        return 'Evening/Sleeper (ES)'
    else:
        return 'Night/Undefined'

# Apply time category
df['TIME_CATEGORY'] = df['SCHEDULED_DEPARTURE'].dt.hour.apply(get_time_category)

# Create GROUP_ID
df['GROUP_ID'] = df['HAUL'] + ' - ' + df['TIME_CATEGORY'].str.split(' ').str[0]

print("✅ Features created")

# Step 3: Create lookup table with percentages
lookup_table = df.groupby('GROUP_ID').agg({
    'TIER1_ELIGIBLE_PAX': 'sum',
    'TIER2_ELIGIBLE_PAX': 'sum',
    'TIER3_ELIGIBLE_PAX': 'sum',
    'TOTAL_PASSENGERS': 'sum'
}).reset_index()

# Calculate percentages
lookup_table['TIER1_PCT'] = (lookup_table['TIER1_ELIGIBLE_PAX'] / lookup_table['TOTAL_PASSENGERS']) * 100
lookup_table['TIER2_PCT'] = (lookup_table['TIER2_ELIGIBLE_PAX'] / lookup_table['TOTAL_PASSENGERS']) * 100
lookup_table['TIER3_PCT'] = (lookup_table['TIER3_ELIGIBLE_PAX'] / lookup_table['TOTAL_PASSENGERS']) * 100

# Keep only necessary columns
lookup_table_final = lookup_table[['GROUP_ID', 'TIER1_PCT', 'TIER2_PCT', 'TIER3_PCT']].round(2)

print("✅ Lookup table created")
print(lookup_table_final)

# Step 4: Merge lookup table WITH main dataframe
df_with_estimates = df.merge(
    lookup_table_final,
    on='GROUP_ID',
    how='left'
)

print(f"✅ Merged successfully. Columns now: {len(df_with_estimates.columns)}")

# Step 5: Calculate estimated demands
df_with_estimates['ESTIMATED_TIER1_DEMAND'] = (
    df_with_estimates['TOTAL_PASSENGERS'] * df_with_estimates['TIER1_PCT'] / 100
).round(0).astype(int)

df_with_estimates['ESTIMATED_TIER2_DEMAND'] = (
    df_with_estimates['TOTAL_PASSENGERS'] * df_with_estimates['TIER2_PCT'] / 100
).round(0).astype(int)

df_with_estimates['ESTIMATED_TIER3_DEMAND'] = (
    df_with_estimates['TOTAL_PASSENGERS'] * df_with_estimates['TIER3_PCT'] / 100
).round(0).astype(int)

# Total estimated demand
df_with_estimates['TOTAL_ESTIMATED_DEMAND'] = (
    df_with_estimates['ESTIMATED_TIER1_DEMAND'] +
    df_with_estimates['ESTIMATED_TIER2_DEMAND'] +
    df_with_estimates['ESTIMATED_TIER3_DEMAND']
)

print("✅ All calculations completed")

# Step 6: Select columns for export
columns_to_export = [
    # Original flight information
    'FLIGHT_DATE',
    'FLIGHT_TIME',
    'TIME_OF_DAY',
    'AIRLINE_CD',
    'FLIGHT_NO',
    'DEPARTURE_STATION_CD',
    'ARRIVAL_STATION_CD',
    'ARRIVAL_COUNTRY',
    'ARRIVAL_REGION',
    'HAUL',
    'AIRCRAFT_TYPE',

    # New engineered features
    'SCHEDULED_DEPARTURE',
    'TIME_CATEGORY',
    'GROUP_ID',

    # Capacity information
    'TOTAL_CAPACITY',
    'TOTAL_PASSENGERS',

    # Lookup table percentages (these came from the merge)
    'TIER1_PCT',
    'TIER2_PCT',
    'TIER3_PCT',

    # Estimated demands
    'ESTIMATED_TIER1_DEMAND',
    'ESTIMATED_TIER2_DEMAND',
    'ESTIMATED_TIER3_DEMAND',
    'TOTAL_ESTIMATED_DEMAND'
]

# Create export dataframe
df_export = df_with_estimates[columns_to_export].copy()

print(f"\n✅ Export dataframe ready: {len(df_export)} rows × {len(df_export.columns)} columns")

# Step 7: Export to Excel
output_file = '/content/drive/My Drive/Colab Notebooks/British airways/BA_Flight_Data_With_Lounge_Estimates.xlsx'

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:

    # Sheet 1: Main flight data with estimates
    df_export.to_excel(writer, sheet_name='Flight Data with Estimates', index=False)

    # Sheet 2: Lookup table reference
    lookup_table_final.to_excel(writer, sheet_name='Lookup Table', index=False)

    # Sheet 3: Summary by GROUP_ID
    summary = df_with_estimates.groupby('GROUP_ID').agg({
        'FLIGHT_NO': 'count',
        'TOTAL_PASSENGERS': 'sum',
        'ESTIMATED_TIER1_DEMAND': 'sum',
        'ESTIMATED_TIER2_DEMAND': 'sum',
        'ESTIMATED_TIER3_DEMAND': 'sum',
        'TOTAL_ESTIMATED_DEMAND': 'sum'
    }).round(0)

    summary.columns = ['Flight Count', 'Total Passengers', 'Total Tier 1', 'Total Tier 2', 'Total Tier 3', 'Total Demand']
    summary.to_excel(writer, sheet_name='Summary by Group')

print("\n" + "="*70)
print("✅ EXCEL FILE EXPORTED SUCCESSFULLY!")
print("="*70)
print(f"\nFile location: {output_file}")
print("\nThe workbook contains 3 sheets:")
print("  1. Flight Data with Estimates - All 10,000 flights with lounge demand")
print("  2. Lookup Table - Reference percentages (6 flight groups)")
print("  3. Summary by Group - Aggregated totals")
print("\n" + "="*70)

# Display sample
print("\n=== SAMPLE OF EXPORTED DATA (First 5 Rows) ===")
print(df_export[['FLIGHT_NO', 'HAUL', 'TIME_CATEGORY', 'GROUP_ID', 'TOTAL_PASSENGERS',
                 'TIER1_PCT', 'ESTIMATED_TIER1_DEMAND',
                 'TIER2_PCT', 'ESTIMATED_TIER2_DEMAND',
                 'TIER3_PCT', 'ESTIMATED_TIER3_DEMAND']].head())

print("\n=== OVERALL TOTALS ===")
print(f"Total Flights: {len(df_export):,}")
print(f"Total Passengers: {df_export['TOTAL_PASSENGERS'].sum():,}")
print(f"Total Tier 1 Demand: {df_export['ESTIMATED_TIER1_DEMAND'].sum():,}")
print(f"Total Tier 2 Demand: {df_export['ESTIMATED_TIER2_DEMAND'].sum():,}")
print(f"Total Tier 3 Demand: {df_export['ESTIMATED_TIER3_DEMAND'].sum():,}")
print(f"Total Lounge Demand: {df_export['TOTAL_ESTIMATED_DEMAND'].sum():,}")
````
